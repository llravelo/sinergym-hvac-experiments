# Sinergym hello world

Verifies the Sinergym + EnergyPlus environment is working. Runs only 10 random steps instead of a full episode so it finishes in seconds. To run a full episode instead, remove the step cap in the loop below (see Sinergym's own `scripts/try_env.py` for that pattern).

```python
import gymnasium as gym
import sinergym

print(sinergym.__version__)
```

```python
env = gym.make('Eplus-5zone-hot-continuous-stochastic-v1')
obs, info = env.reset()
```

```python
for _ in range(10):
    obs, reward, terminated, truncated, info = env.step(env.action_space.sample())
    if terminated or truncated:
        break
```

```python
print(obs, reward)
env.close()
```
