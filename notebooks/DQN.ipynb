# Basic DQN example

Trains a `stable-baselines3` DQN agent on Sinergym's `5ZoneAutoDXVAV` building (`Eplus-5zone-hot-discrete-v1`: discrete action space, `LinearReward` by default). Hyperparameters below are set small on purpose — this is a smoke test that the pipeline (env -> model -> training -> inference) works end to end, not a real training run. A single simulated year is 35,040 timesteps; increase `total_timesteps` and the DQN buffer settings once you want an actual training run.

```python
import gymnasium as gym
import sinergym
from stable_baselines3 import DQN
```

```python
env = gym.make('Eplus-5zone-hot-discrete-v1')
print('action_space:', env.action_space)
print('observation_space:', env.observation_space)
```

```python
model = DQN(
    'MlpPolicy',
    env,
    learning_starts=100,
    buffer_size=1000,
    batch_size=32,
    train_freq=1,
    verbose=1,
)
```

```python
model.learn(total_timesteps=500)
```

```python
obs, info = env.reset()
total_reward = 0
for _ in range(20):
    action, _ = model.predict(obs, deterministic=True)
    obs, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    if terminated or truncated:
        break
print('total_reward:', total_reward)
env.close()
```
